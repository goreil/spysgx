```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import pathlib
import pandas as pd
import numpy as np

OUTPUT_PATH = pathlib.Path(".") / "sub_bytes.csv"
df = pd.read_csv(OUTPUT_PATH)
```

```python
plt.plot(df.loc[0].values[2:])
```

```python
incomplete_data = df.loc[1].values[2:]
n_remove = 10
remove_idx = np.random.choice(incomplete_data.size, n_remove, replace=False)
incomplete_data = np.delete(incomplete_data, remove_idx)

```

```python
fig, ax = plt.subplots(figsize = (15,10))
ax.plot(df.loc[0].values[2:], linestyle  = "--")
ax.plot(incomplete_data, linestyle = "-.")
ax.legend(["Full data", "incomplete data"])

fig.suptitle(f"The missing columns are {remove_idx}")
fig.savefig("missing.pdf")
```
