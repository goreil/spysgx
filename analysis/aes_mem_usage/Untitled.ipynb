```python
import matplotlib.pyplot as plt
```

```python
with open("aes_mem_usage.txt") as f:
    data = [int(x.strip(), 10)//(2**30) for x in f.readlines()]
    
```

```python
with plt.style.context("arend.mplstyle"):
    fig,ax = plt.subplots()
    
    ax.plot(data)
    ax.set_xlabel("Number of steps")
    ax.set_ylabel("Memory usage of process")
    
```
